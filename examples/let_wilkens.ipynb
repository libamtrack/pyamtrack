```python
import pyamtrack.proton_models as pm
import pyamtrack.materials as mat
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
depths = np.linspace(0, 20, 500)
```

```python
results = pm.let_wilkens(depths, 1, 150, 0.01)
```

```python
plt.plot(depths, results)
```

```python
results = pm.let_wilkens(depths, mat.water_liquid, 150, 0.01)
```

```python
plt.plot(depths, results)
```

```python
results = pm.let_wilkens(depths, mat.water_liquid, 150, 0.01, averaging="track")
```

```python
results = pm.let_wilkens(depths, mat.water_liquid, 150, 0.01, averaging=pm.Averaging.TRACK)
```

```python
results = pm.let_wilkens(depths, mat.water_liquid, 150, 0.01, averaging=pm.Averaging.TRACK, cartesian=True)
```
