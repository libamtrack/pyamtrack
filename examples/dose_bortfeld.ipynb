```python
import numpy as np
import matplotlib.pyplot as plt

import pyamtrack

pyamtrack.__version__
```

```python
pm = pyamtrack.proton_models
```

```python
z_cm = np.linspace(0.0, 35.0, 800)

fluence_cm2 = 1e8
E_MeV = 150.0
sigma_E_MeV = -1.0   # libamtrack default = 0.01*E
material = 1         # water
eps = -1.0           # libamtrack default = 0.03

dose = pm.dose_bortfeld(z_cm, fluence_cm2, E_MeV, sigma_E_MeV, material, eps)
dose.shape, float(dose.min()), float(dose.max())
```

```python
plt.figure(figsize=(9, 4))
plt.plot(z_cm, dose, lw=2)
plt.title("Bortfeld 1997: dose vs depth")
plt.xlabel("Depth z [cm]")
plt.ylabel("Dose [Gy]")
plt.grid(alpha=0.3)
plt.show()
```
