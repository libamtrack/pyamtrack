```python
import numpy as np
import matplotlib.pyplot as plt

import pyamtrack

pyamtrack.__version__
```

```python
pm = pyamtrack.proton_models
```

```python
z_cm = np.linspace(0.0, 20.0, 500)

entrance_dose_Gy = 4.0
E_MeV = 150.0
# sigma_E_fraction = 0.01  # 1% energy spread
# material = 1             # water
# eps = 0.03               # nuclear interaction tail fraction

dose = pm.dose_bortfeld(z_cm, entrance_dose_Gy, E_MeV)
dose.shape, float(dose.min()), float(dose.max())
```

```python

plt.figure(figsize=(9, 4))
plt.plot(z_cm, dose, lw=2)
plt.title("Bortfeld 1997: dose vs depth")
plt.xlabel("Depth z [cm]")
plt.ylabel("Dose [Gy]")
plt.grid(alpha=0.3)
plt.show()
```
